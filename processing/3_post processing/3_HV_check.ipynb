## Pemrosesan Data Seismik Pasif, Metode HVSR (*post-processing*)
Author: Annora Vandanu Erlangga

Tahapan selanjutnya setelah pengolahan data seismik pasif dengan metode HVSR adalah pengecekan reliabilitas dan *clear peak* dari kurva HV yang diolah. Pada tahapan ini, sangat dianjurkan untuk mengelola folder penyimpanan data olahan agar tidak membingungkan saat pengolahan lanjutan. Selain itu, pengolahan data seismik pasif juga memerlukan beberapa kali pengolahan hingga mendapatkan hasil yang optimal sehingga akan ada banyak file olahan. Contoh sederhana folder untuk pengolahan HVSR:

```
📁 Processing/
├── 1_merge_mseed/
│   └── [file.mseed]
├── 2_HVSR/
│   └── [file.hv, file.log, file.png]
├── 3_post_processing/
│   └── [script hvsrcheck, kriteria SESAME (2004)]
└── 4_inversi/
    └── [script plotting hasil inversi]
```

Terdapat 2 macam kriteria hasil olahan HVSR, yaitu reliability dan clear peak. Kriteria reliability perlu dipenuhi semua, sedangkan kriteria clear peak perlu minimal 5 dari 6 kriteria dipenuhi. Kriteria tersebut merupakan hasil dari pedoman SESAME yang bertujuan untuk menjamin kualitas dan keandalan interpretasi kurva HVSR, terutama dalam mengidentifikasi frekuensi dominan tanah. Dengan memenuhi kriteria ini, interpretasi dapat dikategorikan ke dalam berbagai bentuk kurva seperti sharp peak, clear peak, unclear low-frequency peak, multiple peaks, broad peak, dan flat curve, yang masing-masing memberikan informasi berbeda mengenai kondisi stratigrafi dan kontras impedansi bawah permukaan.

Kriteria tersebut dihitung dari file `.hv` dan `.log` dari GEOPSY. Penjelasan singkatnya, setelah kalian cek kedua file tersebut, terdapat informasi *Frequency Average*, *Min*, dan *Max*, serta paremeter *windowing*-nya. Nilai-nilai tersebut dapat dianalisis, misalnya dengan Microsoft Excel untuk menghitung standar deviasi kurva. Namun, penggunaan Python dapat mempercepat dan mempermudah proses kalkulasinya. Terdapat modul bernama `hvsrcheck` yang dibuat oleh Aulia Khalqillah, S.Si., M.Si.

#### 1. Instalasi modul `hvsrcheck`
Modul `hvsrcheck` dapat diunduh pada [link berikut](https://github.com/vandanue/hvsrcheck_modified.git). Setelah file `.zip` tersebut diunduh dan **diekstrak** (wajib terutama untuk kalian pengguna Windows 11 yang sering lupa ekstrak file `.zip`), struktur direktori akan tampak seperti ini:

```
📁 hvsrcheck_modified/
├── setup.py
├── geopsy_hvsrcheck.py
├── LICENSE
└── hvcheck/
    ├── __init__.py
    └── hvcheck.py
```

Untuk menginstal modul, buka terminal atau command prompt (CMD), lalu arahkan direktori ke folder hasil ekstraksi (hvsrcheck_modified/), lalu jalankan perintah:

```bash
pip install .
```

Jika instalasi berhasil, kamu akan dapat mengimpor dan menggunakan fungsi dari `hvcheck` atau langsung menggunakan skrip bawaan `geopsy_hvsrcheck.py`.

#### 2. HVSRCHECK
Setelah mengumpulkan semua data olahan HVSR dalam satu folder, langkah selanjutnya adalah menjalankan skrip `geopsy_hvsrcheck.py`. Kalian hanya perlu mengganti folder yang berisikan **semua** file `.hv` dan `.log`.
